```python
import os
import time
import pandas as pd

if os.path.basename(os.getcwd()) == "scripts":
    os.chdir("..")
print("Main Dir:",os.getcwd())
start = time.time()

dir_raw = "data/raw/ieee-fraud-detection"
dir_precessed = "data/processed"

## From our inspection file, we got the required columns name
## Excluding all V and C series columns

txn_columns = ['TransactionID', 
               'isFraud',
               'TransactionDT',
               'TransactionAmt',
               'ProductCD',
               "card1",
               "card2",
               "card3",
               "card4",
               "card5",
               "card6",
               'addr1',
               'addr2',
               'dist1',
               'dist2',
               "P_emaildomain",
               'R_emaildomain']

txn_columns_rename = {
            'TransactionID':"transcation_id", 
            'isFraud':"is_fraud",
            'TransactionDT':"transaction_dt",
            'TransactionAmt':"transaction_amt",
            'ProductCD':"product_cd",
               "card1":"card1",
               "card2":"card2",
               "card3":"card3",
               "card4":"card4",
               "card5":"card5",
               "card6":"card6",
               'addr1':'addr1',
               'addr2':'addr2',
               'dist1':'dist1',
               'dist2':'dist2',
               "P_emaildomain":"p_emaildomain",
               'R_emaildomain':'r_emaildomain'}
def transaction_preprocess():
    print("\n Transaction preprocessing starts....")
    
    input_path = os.path.join(dir_raw,"train_transaction.csv")
    output_path = f"{dir_precessed}/pros_transaction.csv"
    
    ## Read only the needed columns for the Dataframe
    df = pd.read_csv(input_path,usecols=txn_columns)
    print(f"Data Loaded with {df.shape[0]} rows X {df.shape[1]} columns")
    ## Renaming the columns
    df = df.rename(columns=txn_columns_rename)
    ## Rearrange the columns 
    df = df[list(txn_columns_rename.values())]
    
    ## Checking the does the data is loaded fine
    fraud_rate = df["is_fraud"].mean() * 100
    print(f"Fraud rate : {fraud_rate:.2f}%")
    
    ## Exporting to processed folder
    df.to_csv(output_path,na_rep="",index=False)
    elaspsed = time.time() - start
    file_size = os.path.getsize(output_path) / (1024*1024)
    print(f"Saved {output_path} ({file_size:.2f} MB, {elaspsed:.2f}s)")
```

```python
id_columns = [
    'TransactionID',
    'DeviceType',
    'DeviceInfo',
    'id_30',
    'id_31',
    'id_33'
]

id_columns_rename = {
    'TransactionID':  'transaction_id',
    'DeviceType':     'device_type',
    'DeviceInfo':     'device_info',
    'id_30':          'id_30',
    'id_31':          'id_31',
    'id_33':          'id_33'
}

def identity_preprocess():
    print("Identity preprocessing starts.....")
    
    input_path = os.path.join(dir_raw,"train_identity.csv")
    output_path = f'{dir_precessed}/pros_identity.csv'
    
    df = pd.read_csv(input_path,usecols=id_columns)
    print(f"Data Imported with {df.shape[0]} rows X {df.shape[1]} columns ")
    df = df.rename(columns=id_columns_rename)
    df = df[list(id_columns_rename.values())]
    
    print("\n Device type count :")
    print(df['device_type'].value_counts(dropna=False).head().to_string())
    
    df.to_csv(output_path,na_rep="",index=False)
    elaspsed = time.time() - start
    file_size = os.path.getsize(output_path) / (1024*1024)
    print(f"Saved {output_path} ({file_size:.2f} MB, {elaspsed:.2f}s)")
```

```python
transaction_preprocess()
```
